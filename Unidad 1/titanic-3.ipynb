```python
import pandas as pd
 
 
url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/titanic.csv"
df = pd.read_csv(url)
```

SKEUNESS/ASIMETRÍA - Función .skew() 
Esta función calcula la asimetría de una distribución relacionada a su media. Evalúa columnas y filas. Así se aplicaría a la columna Age:

```python
print(df['Age'].skew())
```

--> El resultado es positivo, esto nos indica que según el registro de todas las edades, éstas tienden a los números mâs grandes, lo cual significa que habían más personas grandes que jóvenes.

KURTOSIS - Función .kurt()
Esta función nos permite saber que tan concentrados están los datos alrededor de su media y que tan propensa es una distribucion a presentar valores extremos u outliers.

```python
print(df['Fare'].kurt())
```

--> Como era de esperarse, el resultado es positivo, y con un valor extremadamente elevado de curtosis. Esto es debido a que la columna fare tiene valores muy extremos, el máximo siendo de >500 cuando la mayoría eran tarifas menores de 30. Aquí se demuestran los outliers identificados en la actividad anterior. En este caso las colas de la distribución son muy pesadas mientras que el pico está muy concentrado porque la mayoria de los datos se encuentra ahí, a esto se le conoce como distribución "leptocúrtica".

TABLAS CRUZADAS/CROSS-TABULATION - pd.crosstab()
Esta función muestra cuantos registros caen en cada combinación de datos. 
Se trata de una matriz que muestra la distribución de frecuencias de dos o más variables.

```python
pd.crosstab(df['Pclass'],df['Survived'])
```

Dados los resultados de la tabla cruzada, podemos observar que la clase que tuvo mayor cantidad de sobrevivientes fueron los pasajeros de la primer clase. Los pasajeros de la segunda clase tuvieron cantidades muy similares entre sobrevivientes como no sobrevivientes y los de menor tasa de supervivencia fueron los de la tercera clase.
