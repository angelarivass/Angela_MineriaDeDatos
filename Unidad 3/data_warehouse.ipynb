3. Pregunta: ¿Por qué sería un error catastrófico cargar (Load) ese string gigantesco en la Tabla de Hechos y cómo lo resolverías en tu tubería de datos?

--> Porque la tabla en sí guarda muchos registros (gran cantidad de filas) y el introducir una columna con 2 mb en cada fila hace que el tamaño de la tabla sea innecesariamente gigantesco, y que el tiempo de loading crezca de manera escalada, además de que implica costos más elevados y no aporta al analisis. 
Para resolver este problema decodificaría el Base64 a binario en png para guardarlo en un almacenamiento de archivos (S3/Blob) y cargar en el Datawarehouse únicamente la ruta de la imagen. Así el DW queda ligero y la imagen sigue disponible cuando se necesite.
