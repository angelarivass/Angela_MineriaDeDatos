```python
import pandas as pd
import json
```

```python
with open("raw_actividad_logs.json", encoding="utf-8") as f:
    data = json.load(f)
df_raw = pd.json_normalize(data, sep="_")
```

```python
display(df_raw)
```

```python
df = df_raw.dropna(subset=["account_id", "profile_meta_last_ip", "profile_meta_email_domain"])
df = df[(df['engagement_stats'] != -999) | (df['engagement_stats_avg_session_min'] != -999) | (df['engagement_stats_variance_post_interval_hr'] != -999) | (df['engagement_stats_likes_given_30d'] != -999)]
```

```python
display(df)
```

```python
df['account_id'] = df['account_id'].str.upper()
df = df.drop_duplicates(subset=["account_id"], keep="first")
display(df)
```

```python
df = df.explode("network_activity_retweets_to")
df = df[df['account_id'] != df['network_activity_retweets_to']].reset_index(drop=True)
```

```python
display(df)
```

```python
# Agrupar Interacciones de nuevo
df['network_activity_retweets_to'] = df['network_activity_retweets_to'].str.upper()

final_df = df.groupby(['account_id', 'profile_meta_email_domain', 'profile_meta_last_ip', 'profile_meta_payment_hash', 'engagement_stats_avg_session_min', 'engagement_stats_variance_post_interval_hr', 'engagement_stats_likes_given_30d'], as_index=False, dropna=False)["network_activity_retweets_to"].agg(list)
display(final_df)
```
