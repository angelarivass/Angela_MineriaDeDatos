```python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/angelarivass/Angela_MineriaDeDatos/refs/heads/main/googleplaystore.csv"
 
df = pd.read_csv(url)
```

```python
df.isnull().sum()
```

```python
print("Tamaño del dataset: ", len(df))
```

A: 
Es arquitectónicamente seguro eliminar las filas de Android Ver porque esa cantidad de datos no afecta de manera significativa el análisis o los resultados. 
Por otro lado, si se eliminaran las filas de Rating se estarían eliminando casi 1500 registros de todo el dataset, lo que significa que se perderían una gran cantidad de datos (1,500 de 10,800).

```python
df.dropna(subset=['Android Ver', 'Type'])
```

B:
Como podemos observar, el tamaño del dataset apenas disminuyó:
Antes: 10841
Ahora: 10837

```python
medianas = df.groupby('Category')['Rating'].transform('median')
df['Rating'] = df['Rating'].fillna(medianas)

print("Nulos en Rating:", df['Rating'].isnull().sum())
```

C:
Ya no hay nulos en los ratings.

D: Eliminando valor no numerico en categoría Size, asi se veía antes:

```python
df['Size']
```

Reemplazando "Varies with device" con Nan:

```python
df['Size'] = df['Size'].replace("Varies with device", np.nan)

print(df['Size'])
```

```python
print("Nulos en Size ahora:", df['Size'].isnull().sum())
```
