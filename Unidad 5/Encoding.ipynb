```python
df = pd.read_csv('train_p2.csv')
display(df.head(30))
```

 El país, el dispositivo y el dominio de correo no tienen un "mayor que" o "menor que". Necesitan su propia representación binaria (o de frecuencia) aprendida solo en entrenamiento

```python
#1 

mapeo_verificacion = {
    'NINGUNO': 1,
    'TELEFONO': 2,
    'CORREO': 2,
    'VERIFICADO': 3,
}

df['verfificacion_encoded'] = df['nivel_verificacion'].map(mapeo_verificacion)

print("Resultados del Label Encoding\n")
display(df[['user_id', 'nivel_verificacion', 'verfificacion_encoded']].head())
```
